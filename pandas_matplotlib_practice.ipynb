````python
# 21. Створи DataFrame з кількома пропущеними значеннями (`NaN`) у числових і текстових стовпцях.  
df_1 = pd.DataFrame(
    {
        "Name": ["Anna", "Bohdan", None, "Ira", "Ivan"],
        "Age": [25, np.nan, 30, None, 40],
        "City": ["Kyiv", None, "Lviv", "Odesa", np.nan],
        "Ball": [85.5, 90.0, np.nan, 72.0, None],
        "Course": [1, 3, 5, 4, 4]
    }
)
df_1
````

````python
#22. Замінити всі `NaN` у текстових стовпцях на `"Unknown"_
text_col = df_1.select_dtypes(include=['object']).columns

# Замінюємо NaN на "Unknown" тільки у цих стовпцях
df_1[text_col] = df_1[text_col].fillna("Unknown")

print(df_1)
````

````python
# 23. Замінити `NaN` у числових стовпцях на середнє значення відповідного стовпця.  

numeric_cols = df_1.select_dtypes(include=[np.number]).columns

# 3. Замінюємо NaN у числових стовпцях на середнє значення цього стовпця
df_1[numeric_cols] = df_1[numeric_cols].fillna(df_1[numeric_cols].mean())

print(df_1)
````

````python
# 24. Видалити всі рядки, де хоча б одне значення пропущене.
df_cleaned = df_1.dropna()
df_cleaned
````

````python
# 36. Створи два DataFrame: `customers` і `orders`:
customers_data = {
    'CustomerID': [101, 102, 103, 104, 105],
    'Name': ['Анна', 'Богдан', 'Віктор', 'Галина', 'Дмитро'],
    'City': ['Київ', 'Львів', 'Одеса', 'Київ', 'Харків']
}
customers = pd.DataFrame(customers_data)

# DataFrame ЗАМОВЛЕННЯ (orders)
orders_data = {
    'OrderID': ['01', '02', '03', '04', '05', '06'],
    'CustomerID': [102, 104, 101, 105, 106, 101], # 106 немає в customers
    'Amount': [500.50, 200.00, 150.99, 800.00, 120.00, 300.00]
}
orders = pd.DataFrame(orders_data)

print("--- DataFrame 'customers' ---")
print(customers)
print("\n--- DataFrame 'orders' ---")
print(orders)
````

````python
# 37. Об’єднай дві таблиці вертикально (`pd.concat`).
result_data = pd.concat([customers, orders], axis=0)
print(result_data)
````

````python
# 38. Знайди клієнтів, які **не мають** замовлень (left join + `NaN` у `"Amount"`). 
df_left = pd.merge(
    customers,    # Лівий DataFrame
    orders,    # Правий DataFrame
    on='CustomerID', # Ключ для об'єднання (наприклад, 'CustomerID')
    how='left'        # Тип об'єднання
)
df_left
````

````python
# Фільтруємо ті рядки, де стовпець 'Amount' має пропущене значення (NaN)
customers_not_orders = df_left[df_left["Amount"].isna()] 
customers_not_orders
````

````python
# 39. Збережи результат у CSV-файл (`.to_csv("customers_orders.csv", index=False)`) і потім зчитай його назад (`pd.read_csv`).
customers_not_orders.to_csv("customers_not_orders.csv", index=False)
````

````python
months_labels = j_2025.index  # індекс - це назви місяців, беремо їх для осі X
fig, ax1 = plt.subplots(figsize=(10, 6))

# --- (ax1):---
ax1.bar(months_labels, # будуємо стовчикову діаграму
        j_2025.values, 
        color='blue', 
        label='Revenue 2025')

ax1.set_xlabel('Місяць')
ax1.set_ylabel('Revenue')
ax1.tick_params(axis='y')
ax1.tick_params(axis='x', rotation=45)
# --- (ax2):  ---
ax2 = ax1.twinx()  # Створюємо другу вісь

ax2.plot(months_labels, 
         ave_revenue_2025.values, 
         color="red",
         linewidth=3,
         marker='o',    # Додаємо кружечки
         label='Середнє)')

ax2.set_ylabel('Середнє за 2025')
ax2.tick_params(axis='y')

# 3. Заголовок і оформлення
plt.title('2025: Порівняння продажів із середнім значенням)')
plt.grid(True, axis='x') # Сітка тільки вертикальна, щоб не плутати осі Y


# Забираємо лінії (handles) та підписи (labels) з першої осі — ax1
lines_1, labels_1 = ax1.get_legend_handles_labels()

# Забираємо лінії та підписи з другої осі — ax2 (twinx)
lines_2, labels_2 = ax2.get_legend_handles_labels()

# Об'єднуємо лінії та підписи з обох осей і малюємо одну легенду на ax1
# Це потрібно, тому що при twinx() лінії створюються на різних осях
# і Matplotlib не будує спільну легенду автоматично.
ax1.legend(lines_1 + lines_2, labels_1 + labels_2, loc='upper left')

plt.tight_layout() # Це важливо! Без цього повернуті написи можуть обрізатися знизу
plt.show()
````
