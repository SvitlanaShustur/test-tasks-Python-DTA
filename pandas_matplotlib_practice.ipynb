````python
# 27.Створи DataFrame про продажі з колонками: `"Product"`, `"Category"`, `"Price"`, `"Quantity"` (10–15 рядків).  

sales_data_ua = {
    "Product": [
        "Тональний крем", "Зволожуючий крем", "Туш для вій", "Гігієнічна помада",
        "Сонцезахисний спрей", "Сироватка з вітаміном С", "Рум'яна", "Очищуюча пінка",
        "Тіні для повік", "Бальзам для губ", "Парфуми", "Набір пензлів"
    ],
    "Category": [
        "Макіяж обличчя", "Догляд за шкірою", "Макіяж очей", "Догляд за губами",
        "Спец. догляд", "Догляд за шкірою", "Макіяж обличчя", "Догляд за шкірою",
        "Макіяж очей", "Догляд за губами", "Аксесуари", "Аксесуари"
    ],
    # Ціни з плаваючою комою (приклад, у грн)
    "Price": [
        450.00, 320.50, 210.99, 95.00,
        550.00, 780.00, 150.50, 290.00,
        380.99, 115.00, 1200.00, 420.00
    ],
    # Випадкові цілі числа для кількості
    "Quantity": [
        15, 22, 18, 35,
        10, 8, 25, 30,
        11, 40, 5, 13
    ]
}


df = pd.DataFrame(sales_data_ua)

print(df)
````

````python
# 29. Порахуйте загальний дохід для кожної категорії (`groupby("Category")["Revenue"].sum()`).
df_category = df.groupby("Category")["Revenue"].sum()
df_category
````

````python
# 30. Знайди середню ціну по кожному `"Product"`. 
df_product = df.groupby("Product")["Price"].mean()
df_product
````

````python
# 31. Визнач найпопулярніший товар (за сумарною кількістю продажів). 
df["mode_quantity"] = df['Quantity'].mode()
df
````

````python
# 32. Знайди мінімальний і максимальний дохід по категорії.  
df_Revenue_min = df.groupby("Category")["Revenue"].agg(['min', 'max'])
df_Revenue_min
````

````python
# 33. Зроби зведену таблицю (`pivot_table`) з категоріями по рядках і середнім доходом по стовпцях. 
df_pivot_rev = pd.pivot_table(
    df,
    values='Revenue', #  Значення для агрегації
    index='Category',
    columns='Product',
    aggfunc={'Revenue': 'mean'}, # Агрегуємо по-різному
    fill_value="-"
)
df_pivot_rev
````

````python
# 34. Відсортуй категорії за середнім доходом.
df["Revenue_ave"] = df.groupby("Category")["Revenue"].transform("mean")
df.sort_values("Revenue_ave", inplace=True)
df
````

````python
# Розв'язання для завдань 36–40
# Твій код тут ↓
````

````python
# 36. Створи два DataFrame: `customers` і `orders`:
customers_data = {
    'CustomerID': [101, 102, 103, 104, 105],
    'Name': ['Анна', 'Богдан', 'Віктор', 'Галина', 'Дмитро'],
    'City': ['Київ', 'Львів', 'Одеса', 'Київ', 'Харків']
}
customers = pd.DataFrame(customers_data)

# DataFrame ЗАМОВЛЕННЯ (orders)
orders_data = {
    'OrderID': ['01', '02', '03', '04', '05', '06'],
    'CustomerID': [102, 104, 101, 105, 106, 101], # 106 немає в customers
    'Amount': [500.50, 200.00, 150.99, 800.00, 120.00, 300.00]
}
orders = pd.DataFrame(orders_data)

print("--- DataFrame 'customers' ---")
print(customers)
print("\n--- DataFrame 'orders' ---")
print(orders)
````

````python
# 37. Об’єднай дві таблиці вертикально (`pd.concat`).
result_data = pd.concat([customers, orders], axis=1)
print(result_data)
````

````python
# 38. Знайди клієнтів, які **не мають** замовлень (left join + `NaN` у `"Amount"`). 
df_left = pd.merge(
    customers,    # Лівий DataFrame
    orders,    # Правий DataFrame
    on='CustomerID', # Ключ для об'єднання (наприклад, 'CustomerID')
    how='left'        # Тип об'єднання
)
df_left
````

````python
# Фільтруємо ті рядки, де стовпець 'Amount' має пропущене значення (NaN)
customers_not_orders = df_left[df_left["Amount"].isna()] 
customers_not_orders
````

````python
# 39. Збережи результат у CSV-файл (`.to_csv("customers_orders.csv", index=False)`) і потім зчитай його назад (`pd.read_csv`).
customers_not_orders.to_csv("customers_not_orders.csv", index=False)
````

````python
# зчитай його назад (`pd.read_csv`).
df = pd.read_csv("customers_not_orders.csv")
df
````
